## PnL - ID premium strategy

Baseline: `V(t) = P_max` sold at price `P(t) = P_t`.

ID premium strategy: a share `id_factor` of the volume is valued with an
intraday premium, the remaining share `(1 - id_factor)` stays at price `P(t)`.

- `id_factor` in [0, p%], p = 20%
- `id_premium` = MAX (over the last `x` days, x = 2) of the average (ID3 - DA) per hour bucket
- `id_premium_min` = 0
- `id_premium` clipped to [id_premium_min, 400]

Similar hours (UTC): night 21-05, morning_peak 07-09, solar 10-16,
evening 18-20, shoulder (06 & 17).

```python
#NEW

# ---------------------------------------------------------------------------
# Parameters
# ---------------------------------------------------------------------------
X_DAYS = 2
P_FACTOR = 0.20
PREMIUM_FLOOR = 0
PREMIUM_MAX = 400

ID_COL = "id_vwap_fr"
DA_COL = "da_actual_price"
VOL_COL = "P_max"
PRICE_COL = "P_t"

# France DA vs other countries (for the ID premium min)
FR_DA_COL = "da_storm_fr_fct"
OTHER_DA_COLS = [
    "da_storm_de_fct", "da_storm_be_fct", "da_storm_ch_fct",
    "da_storm_uk_fct", "da_storm_es_fct", "da_storm_it_fct"
]

DEFAULT_HOUR_TO_BUCKET = {
    21: "night", 22: "night", 23: "night", 0: "night", 1: "night", 2: "night", 3: "night", 4: "night", 5: "night",
    7: "morning_peak", 8: "morning_peak", 9: "morning_peak",
    10: "solar", 11: "solar", 12: "solar", 13: "solar", 14: "solar", 15: "solar", 16: "solar",
    18: "evening", 19: "evening", 20: "evening",
    6: "shoulder", 17: "shoulder",
}

def assign_hour_bucket(idx, hour_to_bucket=None):
    """Classify each timestamp into hour buckets."""
    mapping = DEFAULT_HOUR_TO_BUCKET if hour_to_bucket is None else hour_to_bucket
    h = pd.Series(idx.hour, index=idx)
    bucket = h.map(mapping).fillna("other")
    return pd.Series(bucket.values, index=idx, name="hour_bucket")

def build_hour_bucket_map(df_train, n_buckets, spread_col="ID-DA"):
    """
    Build hour->bucket map from train data only:
    group hours by similar mean spread (quantile buckets).
    """
    hourly_spread = df_train.groupby(df_train.index.hour)[spread_col].mean().reindex(range(24))
    if hourly_spread.isna().any():
        # fallback robuste si certaines heures manquent
        hourly_spread = hourly_spread.fillna(hourly_spread.mean())

    # rank -> unique ordering, then qcut into n_buckets groups
    ranked = hourly_spread.rank(method="first")
    labels = [f"bucket_{i}" for i in range(n_buckets)]
    bucket_labels = pd.qcut(ranked, q=n_buckets, labels=labels, duplicates="drop")

    return {int(h): str(bucket_labels.loc[h]) for h in bucket_labels.index}

def compute_pnl(
    df,
    x_days=X_DAYS,
    p_factor=P_FACTOR,
    premium_floor=PREMIUM_FLOOR,
    premium_max=PREMIUM_MAX,
    hour_to_bucket=None,
    resell_remaining_capacity=False,
):
    out = df.copy()

    # --- ID - DA spread and hour buckets -----------------------------------
    out["id_da_spread"] = out["ID-DA"]
    out["hour_bucket"] = assign_hour_bucket(out.index, hour_to_bucket=hour_to_bucket)
    out["day"] = out.index.normalize()

    # --- Raw ID premium -----------------------------------------------------
    daily = out.groupby(["day", "hour_bucket"])["id_da_spread"].mean()
    daily_wide = daily.unstack("hour_bucket").sort_index()
    premium_raw_wide = (
        daily_wide.rolling(window=x_days, min_periods=1)
        .max()
        .shift(1)
    )
    premium_raw = premium_raw_wide.stack().rename("id_premium_raw")
    out = out.join(premium_raw, on=["day", "hour_bucket"])

    # --- Premium final ------------------------------------------------------
    out["id_premium_min"] = float(premium_floor)
    out["id_premium"] = out["id_premium_raw"].clip(
        lower=out["id_premium_min"], upper=premium_max
    )

    # --- Volume split -------------------------------------------------------
    out["id_factor"] = p_factor
    out["vol_da_leg"] = out[VOL_COL] * (1 - out["id_factor"])
    out["vol_id_leg"] = out[VOL_COL] * out["id_factor"]

    # --- Prices used for bids ----------------------------------------------
    out["id_offer_price"] = out["id_premium"].fillna(0.0) + out[PRICE_COL]

    # DA leg accepted if P_t <= DA
    da_accepted = out[PRICE_COL] <= out[DA_COL]
    out["pnl_da_leg_simple"] = np.where(da_accepted, out["vol_da_leg"] * out[DA_COL], 0.0)

    # ID leg accepted if (id_premium + P_t) <= DA
    id_accepted = out["id_offer_price"] <= out[DA_COL]
    out["pnl_id_leg_simple"] = np.where(
        id_accepted, out["vol_id_leg"] * out[DA_COL], 0.0
    )

    # Remaining capacity (not accepted on ID leg)
    out["remaining_capacity"] = np.where(~id_accepted, out["vol_id_leg"], 0.0)

    if resell_remaining_capacity:
        resell_ok = out["id_offer_price"] <= out[ID_COL]
        out["pnl_resell_simple"] = np.where(
            resell_ok, out["remaining_capacity"] * out["id_offer_price"], 0.0
        )
    else:
        out["pnl_resell_simple"] = 0.0

    out["pnl_strategy_simple"] = (
        out["pnl_da_leg_simple"] + out["pnl_id_leg_simple"] + out["pnl_resell_simple"]
    )

    return out[["pnl_strategy_simple"]]
```

```python
#NEW
# --- Grid search sur TRAIN ---
X_DAYS_GRID = [1, 2, 3, 5, 7]
PREMIUM_FLOOR_GRID = [-20, -10, 0, 10, 20]
PREMIUM_MAX_GRID = [150, 250, 400, 600]
N_BUCKETS_GRID = [3, 4, 5, 6, 24]

search_rows = []

for n_buckets in N_BUCKETS_GRID:
    hour_map = build_hour_bucket_map(df_train, n_buckets=n_buckets, spread_col="ID-DA")

    for x_days, p_floor, p_max in product(X_DAYS_GRID, PREMIUM_FLOOR_GRID, PREMIUM_MAX_GRID):
        if p_floor > p_max:
            continue

        pnl_train = compute_pnl(
            df_train,
            x_days=x_days,
            premium_floor=p_floor,
            premium_max=p_max,
            hour_to_bucket=hour_map,
            resell_remaining_capacity=True,
        )["pnl_strategy_simple"].sum()

        search_rows.append({
            "train_pnl": pnl_train,
            "n_buckets": n_buckets,
            "x_days": x_days,
            "premium_floor": p_floor,
            "premium_max": p_max,
            "hour_map": hour_map,
        })

search_df = pd.DataFrame(search_rows).sort_values("train_pnl", ascending=False).reset_index(drop=True)
best = search_df.iloc[0]

best_params = {
    "n_buckets": int(best["n_buckets"]),
    "x_days": int(best["x_days"]),
    "premium_floor": float(best["premium_floor"]),
    "premium_max": float(best["premium_max"]),
}
best_hour_bucket_map = best["hour_map"]

print("Best params (train):", best_params)
print("Best train PnL:", round(float(best["train_pnl"]), 2))
search_df.head(10)
```

```python
# --- Evaluation finale sur TEST avec les meilleurs paramètres ---
df_pnl = pd.DataFrame(index=df_test.index)
df_pnl["pnl_strategy_simple"] = compute_pnl(
    df_test,
    x_days=best_params["x_days"],
    premium_floor=best_params["premium_floor"],
    premium_max=best_params["premium_max"],
    hour_to_bucket=best_hour_bucket_map,
    resell_remaining_capacity=True,
)["pnl_strategy_simple"]

df_pnl
```

## PnL - Benchmark strategy

```python
def compute_base_pnl(
    df,
    vol_col=VOL_COL,
    price_col=PRICE_COL,
    da_col=DA_COL,
    id_col=ID_COL,
    sell_remaining_if_da_le_id=False,
):
    out = df.copy()

    # Acceptation DA : bid price <= DA
    da_accepted = out[price_col] <= out[da_col]

    # PnL base : si accepté, gain = DA * VOL, sinon 0
    out["pnl_base_benchmark"] = np.where(da_accepted, out[da_col] * out[vol_col], 0.0)

    # Volume restant (non vendu en base)
    out["remaining_volume"] = np.where(da_accepted, 0.0, out[vol_col])

    # Option : vendre le volume restant si DA <= ID
    if sell_remaining_if_da_le_id:
        can_sell_remaining = out[price_col] <= out[id_col]
        out["pnl_remaining_benchmark"] = np.where(
            can_sell_remaining,
            out["remaining_volume"] * out[price_col],
            0.0,
        )
    else:
        out["pnl_remaining_benchmark"] = 0.0

    out["pnl_total_benchmark"] = out["pnl_base_benchmark"] + out["pnl_remaining_benchmark"]

    return out["pnl_total_benchmark"]
```

```python
df_pnl["pnl_total_benchmark"] = compute_base_pnl(df_test)
df_pnl
```
